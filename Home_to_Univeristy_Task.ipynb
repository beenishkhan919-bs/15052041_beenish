```python
from collections import deque


# ============================================
# HOME TO UNIVERSITY ROUTE
# BFS AND DFS
# ============================================


# --------------------------------------------
# 1. Create the road map as a graph
# --------------------------------------------

graph = {

    "Shaz Residency": [
        "Hijri Road",
        "Moulana Yousuf Ludhianwi Shaheed Road",
        "Karachi-Hyderabad Motorway"
    ],

    "Hijri Road": [
        "Shaz Residency",
        "Allama Shabbir Ahmed Usmani Road"
    ],

    "Moulana Yousuf Ludhianwi Shaheed Road": [
        "Shaz Residency",
        "Allama Shabbir Ahmed Usmani Road"
    ],

    "Karachi-Hyderabad Motorway": [
        "Shaz Residency",
        "Main University Road"
    ],

    "Allama Shabbir Ahmed Usmani Road": [
        "Hijri Road",
        "Moulana Yousuf Ludhianwi Shaheed Road",
        "Main University Road"
    ],

    "Main University Road": [
        "Allama Shabbir Ahmed Usmani Road",
        "Karachi-Hyderabad Motorway",
        "FUUAST"
    ],

    "FUUAST": [
        "Main University Road"
    ]
}


# ============================================
# 2. Breadth First Search (BFS)
# ============================================

def bfs(graph, start, goal):

    # Queue for BFS
    queue = deque([[start]])

    # Keep track of visited locations
    visited = set()

    while queue:

        # Remove the first path from queue
        path = queue.popleft()

        # Get the current location
        current = path[-1]

        # If already visited, skip it
        if current in visited:
            continue

        # Mark as visited
        visited.add(current)

        # Check if destination is reached
        if current == goal:
            return path

        # Visit neighboring locations
        for neighbor in graph[current]:

            if neighbor not in visited:

                new_path = path + [neighbor]

                queue.append(new_path)

    return None


# ============================================
# 3. Depth First Search (DFS)
# ============================================

def dfs(graph, start, goal):

    # Stack for DFS
    stack = [[start]]

    # Keep track of visited locations
    visited = set()

    while stack:

        # Remove the last path from stack
        path = stack.pop()

        # Get current location
        current = path[-1]

        # If already visited, skip it
        if current in visited:
            continue

        # Mark as visited
        visited.add(current)

        # Check if destination is reached
        if current == goal:
            return path

        # Add neighboring locations to stack
        for neighbor in reversed(graph[current]):

            if neighbor not in visited:

                new_path = path + [neighbor]

                stack.append(new_path)

    return None


# ============================================
# 4. Starting and destination locations
# ============================================

start = "Shaz Residency"

goal = "FUUAST"


# ============================================
# 5. Run BFS
# ============================================

bfs_route = bfs(graph, start, goal)


# ============================================
# 6. Run DFS
# ============================================

dfs_route = dfs(graph, start, goal)


# ============================================
# 7. Display BFS result
# ============================================

print("\n======================================")
print("       BREADTH FIRST SEARCH")
print("======================================")

print("\nBFS Route:")

for i, location in enumerate(bfs_route, 1):

    print(f"{i}. {location}")


# ============================================
# 8. Display DFS result
# ============================================

print("\n======================================")
print("        DEPTH FIRST SEARCH")
print("======================================")

print("\nDFS Route:")

for i, location in enumerate(dfs_route, 1):

    print(f"{i}. {location}")


# ============================================
# 9. Final result
# ============================================

print("\n======================================")
print("              RESULT")
print("======================================")

print("\nBFS:")
print(" -> ".join(bfs_route))

print("\nDFS:")
print(" -> ".join(dfs_route))
```
