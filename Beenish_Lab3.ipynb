```python

!pip install -U google-genai
from google import genai

client = genai.Client(api_key="beenish_apikey")


MODEL = "models/gemini-3.6-flash"


def travel_agent(user_input):

    prompt = f"""
You are a professional travel agent for Pakistan tourism.

Provide a well-structured travel plan with clear headings:

1. Destination Overview
2. Budget Travel Options (Bus, Train, Flight with approx cost in PKR)
3. Places to Visit (specific locations)
4. Food Suggestions (local dishes)
5. Money-Saving Tips

Make response detailed but easy for students.

User request: {user_input}

Travel Plan:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text



print("Gemini Travel Agent Started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye ")
        break

    try:
        reply = travel_agent(user)
        print("\nAgent:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```

```python
!pip install -U google-genai

from google import genai

# Enter your Gemini API Key
client = genai.Client(api_key="Beenish_api_key")


MODEL = "models/gemini-3.6-flash"


def food_delivery_agent(user_input):

    prompt = f"""
You are a professional Food Delivery Agent.

Your job is to help customers choose and order food according to
their preferences, budget, quantity, and requirements.

Provide a well-structured response with clear headings:

1. Food Recommendation
2. Restaurant Suggestions
3. Estimated Price in PKR
4. Delivery Information
5. Special Instructions

Give practical and helpful food suggestions.
If the user provides a budget, make sure the recommendations stay
within that budget as much as possible.

If the user mentions a food preference such as Pakistani, Chinese,
Fast Food, Italian, Vegetarian, Spicy, or Healthy food, recommend
suitable options.

Keep the response detailed, clear, and easy for students to understand.

User request: {user_input}

Food Delivery Plan:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


print("Food Delivery Agent Started (type 'exit' to stop)\n")

while True:

    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye!")
        break

    try:
        reply = food_delivery_agent(user)

        print("\nAgent:\n")
        print(reply)
        print("\n")

    except Exception as e:
        print("Error:", e)
```
