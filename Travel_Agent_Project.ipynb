```python
# ==========================================
# BUDGET AGENT
# ==========================================

class BudgetAgent:

    def __init__(self):

        # Estimated project travel costs in USD
        self.travel_costs = {
            "Dubai": 850,
            "London": 1800,
            "Istanbul": 900,
            "Malaysia": 750,
            "Saudi Arabia": 700,
            "Thailand": 800,
            "Paris": 1600,
            "New York": 2000,
            "Miami": 1900,
            "Toronto": 1700,
            "Singapore": 1000,
            "Bali": 850,
            "Maldives": 1200,
            "Tokyo": 1800,
            "Rome": 1500,
            "Barcelona": 1450,
            "Amsterdam": 1550,
            "Sydney": 2200,
            "Cairo": 800,
            "Baku": 650
        }

    def analyze_budget(self, budget, destination):

        # Find destination without case sensitivity
        selected_destination = None

        for place in self.travel_costs:

            if place.lower() == destination.lower():
                selected_destination = place
                break

        # Destination not found
        if selected_destination is None:
            return {
                "status": "Invalid",
                "message": "Destination is not available in the travel cost database."
            }

        estimated_cost = self.travel_costs[selected_destination]

        # Check affordability
        if budget >= estimated_cost:

            remaining = budget - estimated_cost

            return {
                "status": "Affordable",
                "destination": selected_destination,
                "estimated_cost": estimated_cost,
                "remaining_budget": remaining
            }

        else:

            shortage = estimated_cost - budget

            return {
                "status": "Not Affordable",
                "destination": selected_destination,
                "estimated_cost": estimated_cost,
                "additional_budget_required": shortage
            }


# Create Budget Agent
budget_agent = BudgetAgent()

print("Budget Agent created successfully!")
print("Available destinations:", len(budget_agent.travel_costs))
```

```python
# ==========================================
# SOURCE AGENT - API VERSION
# ==========================================

import requests


class SourceAgent:

    def __init__(self):
        self.api_url = "https://geocoding-api.open-meteo.com/v1/search"

    def find_source(self, source):

        params = {
            "name": source,
            "count": 1,
            "language": "en",
            "format": "json"
        }

        try:
            response = requests.get(self.api_url, params=params, timeout=10)

            if response.status_code != 200:
                return {
                    "status": "Error",
                    "message": "Unable to connect to location API."
                }

            data = response.json()

            if "results" not in data or len(data["results"]) == 0:
                return {
                    "status": "Invalid",
                    "message": "Source location was not found."
                }

            location = data["results"][0]

            return {
                "status": "Valid",
                "city": location.get("name"),
                "country": location.get("country"),
                "latitude": location.get("latitude"),
                "longitude": location.get("longitude"),
                "timezone": location.get("timezone")
            }

        except requests.exceptions.RequestException:
            return {
                "status": "Error",
                "message": "Network error while connecting to API."
            }


# Create Source Agent
source_agent = SourceAgent()

print("Source Agent created successfully!")
print("API connection configured.")
```

```python
# ==========================================
# DESTINATION AGENT - API VERSION
# ==========================================

import requests


class DestinationAgent:

    def __init__(self):
        self.api_url = "https://geocoding-api.open-meteo.com/v1/search"

    def find_destination(self, destination):

        params = {
            "name": destination,
            "count": 1,
            "language": "en",
            "format": "json"
        }

        try:
            response = requests.get(
                self.api_url,
                params=params,
                timeout=10
            )

            if response.status_code != 200:
                return {
                    "status": "Error",
                    "message": "Unable to connect to destination API."
                }

            data = response.json()

            if "results" not in data or len(data["results"]) == 0:
                return {
                    "status": "Invalid",
                    "message": "Destination was not found."
                }

            location = data["results"][0]

            return {
                "status": "Valid",
                "destination": location.get("name"),
                "country": location.get("country"),
                "latitude": location.get("latitude"),
                "longitude": location.get("longitude"),
                "timezone": location.get("timezone")
            }

        except requests.exceptions.RequestException:
            return {
                "status": "Error",
                "message": "Network error while connecting to API."
            }


# Create Destination Agent
destination_agent = DestinationAgent()

print("Destination Agent created successfully!")
print("API connection configured.")
```

```python
# ==========================================
# FINAL AI TRAVEL AGENT
# CONNECTS ALL THREE AGENTS
# ==========================================

class AITravelAgent:

    def __init__(self, budget_agent, source_agent, destination_agent):

        self.budget_agent = budget_agent
        self.source_agent = source_agent
        self.destination_agent = destination_agent

    def plan_trip(self, budget, source, destination):

        # ------------------------------------------
        # 1. SOURCE AGENT
        # ------------------------------------------

        source_result = self.source_agent.find_source(source)

        # ------------------------------------------
        # 2. DESTINATION AGENT
        # ------------------------------------------

        destination_result = self.destination_agent.find_destination(destination)

        # ------------------------------------------
        # 3. BUDGET AGENT
        # ------------------------------------------

        if destination_result["status"] == "Valid":

            budget_result = self.budget_agent.analyze_budget(
                budget,
                destination_result["destination"]
            )

        else:

            budget_result = {
                "status": "Skipped",
                "message": "Budget analysis skipped because destination is invalid."
            }

        # ------------------------------------------
        # 4. FINAL DECISION
        # ------------------------------------------

        if (
            source_result["status"] == "Valid"
            and destination_result["status"] == "Valid"
            and budget_result["status"] == "Affordable"
        ):

            decision = "TRIP APPROVED"

        else:

            decision = "TRIP NOT APPROVED"

        return {
            "source": source_result,
            "destination": destination_result,
            "budget": budget_result,
            "decision": decision
        }


# Create the complete AI Travel Agent
travel_agent = AITravelAgent(
    budget_agent,
    source_agent,
    destination_agent
)

print("AI Travel Agent created successfully!")
print("All three agents are now connected.")
```

```python
# ==========================================
# TEST THE COMPLETE AI TRAVEL AGENT
# ==========================================

print("=" * 60)
print("              AI TRAVEL AGENT")
print("=" * 60)

budget = float(input("Enter your budget (USD): "))
source = input("Enter your source city: ").strip()
destination = input("Enter your destination: ").strip()

result = travel_agent.plan_trip(
    budget,
    source,
    destination
)

print("\n" + "=" * 60)
print("                 TRAVEL PLAN")
print("=" * 60)

# ------------------------------------------
# SOURCE AGENT RESULT
# ------------------------------------------

print("\n--- SOURCE AGENT ---")

if result["source"]["status"] == "Valid":

    print("Status:", result["source"]["status"])
    print("City:", result["source"]["city"])
    print("Country:", result["source"]["country"])
    print("Latitude:", result["source"]["latitude"])
    print("Longitude:", result["source"]["longitude"])
    print("Timezone:", result["source"]["timezone"])

else:

    print("Status:", result["source"]["status"])
    print("Message:", result["source"]["message"])


# ------------------------------------------
# DESTINATION AGENT RESULT
# ------------------------------------------

print("\n--- DESTINATION AGENT ---")

if result["destination"]["status"] == "Valid":

    print("Status:", result["destination"]["status"])
    print("Destination:", result["destination"]["destination"])
    print("Country:", result["destination"]["country"])
    print("Latitude:", result["destination"]["latitude"])
    print("Longitude:", result["destination"]["longitude"])
    print("Timezone:", result["destination"]["timezone"])

else:

    print("Status:", result["destination"]["status"])
    print("Message:", result["destination"]["message"])


# ------------------------------------------
# BUDGET AGENT RESULT
# ------------------------------------------

print("\n--- BUDGET AGENT ---")

if result["budget"]["status"] == "Affordable":

    print("Status:", result["budget"]["status"])
    print("Estimated Cost: $", result["budget"]["estimated_cost"])
    print("Remaining Budget: $", result["budget"]["remaining_budget"])

elif result["budget"]["status"] == "Not Affordable":

    print("Status:", result["budget"]["status"])
    print("Estimated Cost: $", result["budget"]["estimated_cost"])
    print(
        "Additional Budget Required: $",
        result["budget"]["additional_budget_required"]
    )

else:

    print("Status:", result["budget"]["status"])
    print("Message:", result["budget"]["message"])


# ------------------------------------------
# FINAL DECISION
# ------------------------------------------

print("\n" + "=" * 60)
print("                FINAL DECISION")
print("=" * 60)

if result["decision"] == "TRIP APPROVED":

    print("✅", result["decision"])
    print("Your trip is possible within the estimated budget.")

else:

    print("❌", result["decision"])

    if result["source"]["status"] != "Valid":
        print("• Source location could not be validated.")

    if result["destination"]["status"] != "Valid":
        print("• Destination could not be validated.")

    if result["budget"]["status"] == "Not Affordable":
        print("• Budget is not sufficient for this destination.")

print("=" * 60)
```
